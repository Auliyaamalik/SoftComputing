```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

x = np.arange(0, 101, 1)
pad = 10

A = fuzz.trapmf(x, [86 - pad, 86, 100, 100 + pad])
B = fuzz.trapmf(x, [66 - pad, 66, 75, 75 + pad])
C = fuzz.trapmf(x, [56 - pad, 56, 60, 60 + pad])

plt.figure(figsize=(10, 4.8))

# Fill (warna di dalam)
plt.fill_between(x, 0, C, color="tab:blue", alpha=0.25, label="C (filled)")
plt.fill_between(x, 0, B, color="tab:orange", alpha=0.25, label="B (filled)")
plt.fill_between(x, 0, A, color="tab:green", alpha=0.25, label="A (filled)")

# Garis (biar jelas)
plt.plot(x, C, color="tab:blue", linewidth=2.5, label="C")
plt.plot(x, B, color="tab:orange", linewidth=2.5, label="B")
plt.plot(x, A, color="tab:green", linewidth=2.5, label="A")

plt.title("Fuzzy Membership Function (TrapMF) - Filled")
plt.xlabel("Nilai (0-100)")
plt.ylabel("Derajat Keanggotaan (0-1)")
plt.xlim(0, 100)
plt.ylim(-0.02, 1.05)
plt.grid(True, alpha=0.25)

# Biar legend nggak dobel (karena ada filled + line)
handles, labels = plt.gca().get_legend_handles_labels()
seen = set()
uniq_h, uniq_l = [], []
for h, l in zip(handles, labels):
    if l not in seen and l in ["A", "B", "C"]:
        uniq_h.append(h); uniq_l.append(l); seen.add(l)
plt.legend(uniq_h, uniq_l, loc="upper left")

plt.tight_layout()
plt.show()
```
