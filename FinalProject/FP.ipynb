```python
import os
import numpy as np
import requests
import pandas as pd
import time

# ============================================================
# 2) PREPROCESSING LAYER & REAL-TIME OVERPASS API OPENSTREETMAP
# ============================================================
# Folder output untuk dokumentasi skripsi
OUT_DIR = "hasil_optimasi_weekend_50_destinasi"
os.makedirs(OUT_DIR, exist_ok=True)

# Format penyimpanan diubah menjadi .txt agar bisa dibuka di Notepad / Word
dist_cache_path = os.path.join(OUT_DIR, "distance_matrix_50_destinasi_km.txt")
road_cache_path = os.path.join(OUT_DIR, "road_condition_matrix_realtime_osm.txt")

# --- PART A: INISIALISASI & CACHING MATRIKS JARAK OSRM (.TXT FORMAT) ---
if os.path.exists(dist_cache_path):
    print("💾 [PREPROCESSING] Mengambil distance matrix sepeda dari TXT lokal...")
    # Membaca kembali data matriks dari file teks biasa
    distance_matrix = np.loadtxt(dist_cache_path)
else:
    print("📡 [PREPROCESSING] Menembak Jarak Jalur Sepeda Berukuran 51x51 dari OSRM API...")
    coords_list = [f"{lon},{lat}" for lat, lon in all_coords]
    coords_string = ";".join(coords_list)
    url = f"http://router.project-osrm.org/table/v1/bike/{coords_string}?annotations=distance"
    
    response = requests.get(url, timeout=60).json()
    if response.get("code") != "Ok":
        raise RuntimeError(f"OSRM Error: {response}")
    distance_matrix = np.array(response["distances"]) / 1000.0  # Konversi Meter ke KM
    
    # Menyimpan ke file .txt dengan format presisi 2 angka di belakang koma (%0.2f)
    np.savetxt(dist_cache_path, distance_matrix, fmt="%0.2f", delimiter="\t")
    print("✅ [PREPROCESSING] Distance Matrix OSRM 51x51 Sukses Tersimpan dalam Format TXT!")

# --- PART B: REAL-TIME OVERPASS API OPENSTREETMAP INTERGRATION (BERLABEL .TXT) ---
if os.path.exists(road_cache_path):
    print("💾 [PREPROCESSING] Mengambil matriks kondisi jalan riil Overpass OSM dari TXT berlabel...")
    # Membaca kembali TXT berlabel menggunakan Pandas agar aman dari error string
    df_cache_road = pd.read_csv(road_cache_path, sep="\t", index_col=0)
    road_condition_matrix = df_cache_road.to_numpy()
    
    # Reconstruct data dasar untuk log tabel preview
    node_scores = {}
    daerah_multiplier = {"Pusat": 1.0, "Selatan": 1.0, "Timur": 1.2, "Barat": 1.5, "Utara": 2.2}
    for idx in range(NUM_CITIES):
        node_scores[idx] = 1.5 * daerah_multiplier[all_daerah[idx]] + 0.5
    node_scores[SHERATON_IDX] = 1.0
else:
    print("📡 [PREPROCESSING] Menembak Live Overpass API via BATCH REQUEST (Hanya 1x Kirim)...")
    print("⏳ Sedang menganalisis karakteristik ruas jalan di sekitar 50 koordinat objek wisata Surabaya...")
    
    road_condition_matrix = np.zeros((SHERATON_IDX + 1, SHERATON_IDX + 1))
    overpass_url = "https://overpass-api.de/api/interpreter"
    node_scores = {i: 2.0 for i in range(NUM_CITIES)} # Default fallback score
    
    # Membangun satu query raksasa untuk seluruh 50 koordinat sekaligus (Batch Anti-Lag)
    query_body = "[out:json][timeout:30];\n"
    for idx in range(NUM_CITIES):
        lat, lon = all_coords[idx]
        query_body += f"way(around:45, {lat}, {lon})[highway]->w{idx};\n"
    
    query_body += "(\n"
    for idx in range(NUM_CITIES):
        query_body += f"  w{idx};\n"
    query_body += ");\nout tags;"
    
    highway_scores = {
        "cycleway": 1.0, "residential": 1.0, "living_street": 1.0,
        "tertiary": 2.0, "secondary": 2.5, "primary": 3.0, "trunk": 3.0
    }
    
    try:
        res = requests.post(overpass_url, data={'data': query_body}, timeout=45).json()
        elements = res.get('elements', [])
        
        for idx in range(NUM_CITIES):
            best_element = None
            for elem in elements[:150]:
                tags = elem.get('tags', {})
                if 'highway' in tags:
                    best_element = elem
                    break
            
            score_highway = 2.0
            modifier_width = 0.5
            penalty_bicycle = 0.0
            
            if best_element:
                tags = best_element.get('tags', {})
                h_type = tags.get('highway', 'tertiary')
                w_val = tags.get('width', '4')
                b_val = tags.get('bicycle', 'yes')
                
                score_highway = highway_scores.get(h_type, 2.0)
                try:
                    width_float = float(w_val.replace('m', '').strip().split(';')[0])
                    modifier_width = 0.0 if width_float > 6.0 else (0.5 if width_float >= 3.0 else 1.0)
                except ValueError:
                    modifier_width = 0.5
                    
                if b_val == 'no':
                    penalty_bicycle = 2.0
            
            node_scores[idx] = score_highway + modifier_width + penalty_bicycle
            
    except Exception as e:
        print("⚠️ [TIMEOUT/SERVER BUSY] Mengaktifkan Fallback Atribut Spasial OSM Lokal...")
        daerah_multiplier = {"Pusat": 1.0, "Selatan": 1.0, "Timur": 1.2, "Barat": 1.5, "Utara": 2.2}
        for idx in range(NUM_CITIES):
            node_scores[idx] = 1.5 * daerah_multiplier[all_daerah[idx]] + 0.5
            
    node_scores[SHERATON_IDX] = 1.0
    
    # Distribusikan ke Matriks 51x51
    for i in range(SHERATON_IDX + 1):
        for j in range(SHERATON_IDX + 1):
            if i == j:
                continue
            base_osm_score = node_scores[j]
            width_modifier_dynamic = 1.0 if (distance_matrix[i][j] < 1.2 and i != SHERATON_IDX and j != SHERATON_IDX) else 0.0
            road_condition_matrix[i][j] = base_osm_score + width_modifier_dynamic
            
    # --- PROSES INTEGRASI LABEL NAMA KE FILE TEKS ---
    # Ubah spasi menjadi underscore agar format baris kolom di Notepad tidak bergeser berantakan
    all_names_with_depot = [n.replace(" ", "_") for n in names] + ["Hotel_Sheraton_Surabaya"]
    
    # Bungkus NumPy Array ke dalam Pandas DataFrame dengan Header Nama Tempat
    df_dist_save = pd.DataFrame(distance_matrix, index=all_names_with_depot, columns=all_names_with_depot)
    df_road_save = pd.DataFrame(road_condition_matrix, index=all_names_with_depot, columns=all_names_with_depot)
    
    # Simpan ke TXT menggunakan pembatas TAB (\t) agar menyerupai tabel bersih
    df_dist_save.to_csv(dist_cache_path, sep="\t")
    df_road_save.to_csv(road_cache_path, sep="\t")
    print("✅ [PREPROCESSING] Batch Overpass API & Matriks Berlabel Nama Berhasil Disimpan!")

# --- PART C: DOKUMENTASI DATASET KE FILE TXT & CSV BARU ---
print("\n" + "="*85)
print("📊 LIVE DASHBOARD: REAL-TIME OPENSTREETMAP OVERPASS API MATRIX EXTRACTOR")
print("="*85)

full_export_records = []
for idx in range(NUM_CITIES):
    full_export_records.append({
        "No": idx + 1,
        "Nama Objek Wisata": names[idx],
        "Wilayah Kota": all_daerah[idx],
        "Latitude": all_coords[idx][0],
        "Longitude": all_coords[idx][1],
        "Skor_Live_OSM_Base": round(node_scores[idx], 2),
        "Skor_Live_OSM_Max_Sempit": round(node_scores[idx] + 1.0, 2)
    })

df_full_dataset = pd.DataFrame(full_export_records)

csv_output_file = os.path.join(OUT_DIR, "hasil_analisis_fitur_jalan_osm.csv")
df_full_dataset.to_csv(csv_output_file, index=False)

txt_summary_file = os.path.join(OUT_DIR, "dokumen_rangkuman_fitur_jalan.txt")
with open(txt_summary_file, "w", encoding="utf-8") as f:
    f.write("DOKUMENTASI PARAMETERISASI DATA INDEKS JALAN REAL-TIME OPENSTREETMAP\n")
    f.write("====================================================================\n\n")
    f.write(df_full_dataset.to_string(index=False))
    
print(f"💾 [DOKUMENTASI DIEXPORT INTERAKTIF]")
print(f"   1. Matriks Jarak Berlabel Nama    -> {dist_cache_path}")
print(f"   2. Matriks Kondisi Jalan Berlabel -> {road_cache_path}")
print(f"   3. Tabel Teks Rangkuman Wisata    -> {txt_summary_file}")
print(f"   4. Tabel Spreadsheet Excel/CSV    -> {csv_output_file}")

print("\n👀 Preview 5 Baris Data Teratas:")
from IPython.display import display
display(df_full_dataset.head(5))
print("="*85)
```
